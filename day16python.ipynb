```python
# ============================================================
# STUDENT WELLBEING SURVEY
# COMPLETE STATISTICAL ANALYSIS
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES AND UPLOAD DATASET
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

# Google Colab file upload
from google.colab import files

print("=" * 70)
print("STUDENT WELLBEING SURVEY - STATISTICAL ANALYSIS")
print("=" * 70)

print("\nPlease upload the Student Wellbeing Survey CSV file.")

uploaded = files.upload()

# Automatically get the uploaded filename
file_path = list(uploaded.keys())[0]

# Read the dataset
df = pd.read_csv(file_path)

print("\nDataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nNumber of Students:")
print(len(df))

print("\nFirst 5 Rows:")
display(df.head())


# ============================================================
# 2. INSPECT THE DATASET
# ============================================================

print("\n" + "=" * 70)
print("2. DATASET INSPECTION")
print("=" * 70)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))

print("\nMissing Values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDataset Information:")
df.info()


# ============================================================
# 3. MEAN, MEDIAN AND MODE
# ============================================================

print("\n" + "=" * 70)
print("3. MEAN, MEDIAN AND MODE")
print("=" * 70)

variables = [
    "Weekly_Study_Hours",
    "Average_Sleep_Hours",
    "Daily_Screen_Time_Hours",
    "Stress_Score",
    "Academic_Readiness_Score"
]

central_statistics = []

for column in variables:

    mean_value = df[column].mean()

    median_value = df[column].median()

    mode_values = df[column].mode()

    if len(mode_values) > 0:
        mode_value = ", ".join(str(x) for x in mode_values)
    else:
        mode_value = "No mode"

    central_statistics.append({
        "Variable": column,
        "Mean": mean_value,
        "Median": median_value,
        "Mode": mode_value
    })

central_df = pd.DataFrame(central_statistics)

print("\nMean, Median and Mode:")
display(central_df.round(4))


# ============================================================
# 4. RANGE, VARIANCE, STANDARD DEVIATION, Q1, Q3 AND IQR
# ============================================================

print("\n" + "=" * 70)
print("4. MEASURES OF VARIABILITY")
print("=" * 70)

spread_statistics = []

for column in variables:

    minimum = df[column].min()

    maximum = df[column].max()

    range_value = maximum - minimum

    variance_value = df[column].var()

    standard_deviation = df[column].std()

    q1 = df[column].quantile(0.25)

    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    spread_statistics.append({
        "Variable": column,
        "Minimum": minimum,
        "Maximum": maximum,
        "Range": range_value,
        "Variance": variance_value,
        "Standard Deviation": standard_deviation,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr
    })

spread_df = pd.DataFrame(spread_statistics)

print("\nRange, Variance, Standard Deviation, Q1, Q3 and IQR:")
display(spread_df.round(4))


# ============================================================
# 5. IDENTIFY VARIABLE WITH GREATEST VARIABILITY
# ============================================================

print("\n" + "=" * 70)
print("5. GREATEST VARIABILITY")
print("=" * 70)

greatest_std_variable = spread_df.loc[
    spread_df["Standard Deviation"].idxmax(),
    "Variable"
]

greatest_range_variable = spread_df.loc[
    spread_df["Range"].idxmax(),
    "Variable"
]

greatest_iqr_variable = spread_df.loc[
    spread_df["IQR"].idxmax(),
    "Variable"
]

greatest_std_value = spread_df.loc[
    spread_df["Standard Deviation"].idxmax(),
    "Standard Deviation"
]

print(
    "Variable with greatest variability based on Standard Deviation:",
    greatest_std_variable
)

print(
    "Variable with greatest Range:",
    greatest_range_variable
)

print(
    "Variable with greatest IQR:",
    greatest_iqr_variable
)

print(
    "Greatest Standard Deviation:",
    round(greatest_std_value, 4)
)

print("\nInterpretation:")

print(
    f"{greatest_std_variable} has the greatest variability "
    f"because it has the highest standard deviation "
    f"({greatest_std_value:.4f}) among the five variables."
)


# ============================================================
# 6. IQR OUTLIER DETECTION
# ============================================================

print("\n" + "=" * 70)
print("6. IQR OUTLIER DETECTION")
print("=" * 70)

outlier_variables = [
    "Weekly_Study_Hours",
    "Daily_Screen_Time_Hours",
    "Commute_Time_Minutes",
    "Monthly_Discretionary_Spending"
]

outlier_results = {}

outlier_summary = []

for column in outlier_variables:

    q1 = df[column].quantile(0.25)

    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)

    upper_bound = q3 + (1.5 * iqr)

    outlier_condition = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    )

    outliers = df.loc[
        outlier_condition,
        column
    ]

    outlier_results[column] = outlier_condition

    outlier_summary.append({
        "Variable": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Number of Outliers": len(outliers),
        "Outlier Values": outliers.tolist()
    })

outlier_df = pd.DataFrame(outlier_summary)

print("\nIQR Outlier Results:")
display(outlier_df.round(4))


# ============================================================
# 7. DISPLAY OUTLIER VALUES
# ============================================================

print("\n" + "=" * 70)
print("7. OUTLIER VALUES")
print("=" * 70)

for column in outlier_variables:

    condition = outlier_results[column]

    outliers = df.loc[
        condition,
        column
    ]

    print("\nVariable:", column)

    print("Number of outliers:", len(outliers))

    if len(outliers) > 0:

        print("Outlier values:")
        print(outliers.tolist())

    else:

        print("No outliers detected.")


# ============================================================
# 8. BEFORE AND AFTER OUTLIER REMOVAL
# ============================================================

print("\n" + "=" * 70)
print("8. BEFORE VS AFTER OUTLIER REMOVAL")
print("=" * 70)

column = "Weekly_Study_Hours"

outlier_condition = outlier_results[column]

# Before removing outliers
before_mean = df[column].mean()

before_median = df[column].median()

# Remove outliers
clean_data = df.loc[
    ~outlier_condition,
    column
]

# After removing outliers
after_mean = clean_data.mean()

after_median = clean_data.median()

comparison_df = pd.DataFrame({
    "Statistic": ["Mean", "Median"],
    "Before Outlier Removal": [
        before_mean,
        before_median
    ],
    "After Outlier Removal": [
        after_mean,
        after_median
    ],
    "Change": [
        after_mean - before_mean,
        after_median - before_median
    ]
})

print("\nWeekly Study Hours:")
display(comparison_df.round(4))

print(
    f"\nBefore removing outliers, the mean was "
    f"{before_mean:.4f} and the median was "
    f"{before_median:.4f}."
)

print(
    f"After removing outliers, the mean was "
    f"{after_mean:.4f} and the median was "
    f"{after_median:.4f}."
)

print("\nInterpretation:")

print(
    "The comparison shows how extreme observations can influence "
    "the mean. The median is generally more resistant to outliers."
)


# ============================================================
# 9. DEFINE PROBABILITY EVENTS
# ============================================================

print("\n" + "=" * 70)
print("9. PROBABILITY ANALYSIS")
print("=" * 70)

# Event A:
# Student has a Part-Time Job = Yes

A = (
    df["Part_Time_Job"]
    .astype(str)
    .str.strip()
    .str.lower()
    == "yes"
)

# Event B:
# Stress Score >= 7

B = df["Stress_Score"] >= 7

# Event C:
# Scholarship = Yes

C = (
    df["Scholarship"]
    .astype(str)
    .str.strip()
    .str.lower()
    == "yes"
)

# Event D:
# Exercises at least 3 days per week

D = df["Exercise_Days_Per_Week"] >= 3

n = len(df)

print("Total number of students:", n)

print("\nEvent A: Part_Time_Job = Yes")
print("Students satisfying A:", A.sum())

print("\nEvent B: Stress_Score >= 7")
print("Students satisfying B:", B.sum())

print("\nEvent C: Scholarship = Yes")
print("Students satisfying C:", C.sum())

print("\nEvent D: Exercise_Days_Per_Week >= 3")
print("Students satisfying D:", D.sum())


# ============================================================
# 10. BASIC PROBABILITIES
# ============================================================

print("\n" + "=" * 70)
print("10. BASIC PROBABILITIES")
print("=" * 70)

# Formula:
# P(Event) = Number of favorable outcomes / Total outcomes

P_A = A.mean()

P_B = B.mean()

P_C = C.mean()

P_D = D.mean()

print("Formula:")
print(
    "P(Event) = Number of favorable outcomes / "
    "Total number of students"
)

print("\nP(A):")

print(
    f"{A.sum()} / {n} = {P_A:.6f}"
)

print("\nP(B):")

print(
    f"{B.sum()} / {n} = {P_B:.6f}"
)

print("\nP(C):")

print(
    f"{C.sum()} / {n} = {P_C:.6f}"
)

print("\nP(D):")

print(
    f"{D.sum()} / {n} = {P_D:.6f}"
)


# ============================================================
# 11. P(A OR B)
# ============================================================

print("\n" + "=" * 70)
print("11. P(A OR B)")
print("=" * 70)

# Formula:
# P(A or B) = P(A) + P(B) - P(A and B)

A_or_B = A | B

P_A_and_B = (A & B).mean()

P_A_or_B = A_or_B.mean()

print("Formula:")

print(
    "P(A or B) = P(A) + P(B) - P(A and B)"
)

print(
    f"\nP(A or B) = "
    f"{P_A:.6f} + {P_B:.6f} - {P_A_and_B:.6f}"
)

print(
    f"P(A or B) = {P_A_or_B:.6f}"
)

print(
    "\nNumber of students satisfying A or B:",
    A_or_B.sum()
)


# ============================================================
# 12. P(A AND B)
# ============================================================

print("\n" + "=" * 70)
print("12. P(A AND B)")
print("=" * 70)

A_and_B = A & B

P_A_and_B = A_and_B.mean()

print("Formula:")

print(
    "P(A and B) = Number of students satisfying "
    "both A and B / Total"
)

print(
    f"\nP(A and B) = {A_and_B.sum()} / {n}"
)

print(
    f"P(A and B) = {P_A_and_B:.6f}"
)


# ============================================================
# 13. CONDITIONAL PROBABILITIES
# ============================================================

print("\n" + "=" * 70)
print("13. CONDITIONAL PROBABILITIES")
print("=" * 70)

# Formula:
# P(A|B) = P(A and B) / P(B)

P_A_given_B = P_A_and_B / P_B

# Formula:
# P(B|A) = P(A and B) / P(A)

P_B_given_A = P_A_and_B / P_A

print("Formula:")

print(
    "P(A|B) = P(A and B) / P(B)"
)

print(
    f"\nP(A|B) = "
    f"{P_A_and_B:.6f} / {P_B:.6f}"
)

print(
    f"P(A|B) = {P_A_given_B:.6f}"
)

print("\nFormula:")

print(
    "P(B|A) = P(A and B) / P(A)"
)

print(
    f"\nP(B|A) = "
    f"{P_A_and_B:.6f} / {P_A:.6f}"
)

print(
    f"P(B|A) = {P_B_given_A:.6f}"
)

print(
    f"\nInterpretation: Among students with Stress Score "
    f">= 7, {P_A_given_B * 100:.2f}% have a part-time job."
)

print(
    f"Among students with a part-time job, "
    f"{P_B_given_A * 100:.2f}% have Stress Score >= 7."
)


# ============================================================
# 14. PROBABILITY SUMMARY TABLE
# ============================================================

print("\n" + "=" * 70)
print("14. PROBABILITY SUMMARY")
print("=" * 70)

probability_summary = pd.DataFrame({
    "Probability": [
        "P(A)",
        "P(B)",
        "P(C)",
        "P(D)",
        "P(A or B)",
        "P(A and B)",
        "P(A|B)",
        "P(B|A)"
    ],

    "Value": [
        P_A,
        P_B,
        P_C,
        P_D,
        P_A_or_B,
        P_A_and_B,
        P_A_given_B,
        P_B_given_A
    ]
})

probability_summary["Percentage"] = (
    probability_summary["Value"] * 100
)

display(
    probability_summary.round(6)
)


# ============================================================
# 15. MUTUALLY EXCLUSIVE EVENTS
# ============================================================

print("\n" + "=" * 70)
print("15. MUTUALLY EXCLUSIVE EVENTS")
print("=" * 70)

Year_1 = df["Year_of_Study"] == 1

Year_4 = df["Year_of_Study"] == 4

intersection = Year_1 & Year_4

intersection_count = intersection.sum()

P_intersection = intersection.mean()

print(
    "Event Y1: Year_of_Study = 1"
)

print(
    "Event Y4: Year_of_Study = 4"
)

print(
    "\nNumber of students in both Year 1 and Year 4:",
    intersection_count
)

print(
    "P(Year 1 and Year 4):",
    P_intersection
)

if intersection_count == 0:

    print(
        "\nConclusion: Year 1 and Year 4 are "
        "MUTUALLY EXCLUSIVE."
    )

else:

    print(
        "\nConclusion: Year 1 and Year 4 are "
        "NOT mutually exclusive."
    )


# ============================================================
# 16. TEST WHETHER A AND B ARE INDEPENDENT
# ============================================================

print("\n" + "=" * 70)
print("16. INDEPENDENCE OF A AND B")
print("=" * 70)

# Formula:
#
# If A and B are independent:
#
# P(A and B) = P(A) × P(B)

product = P_A * P_B

difference = P_A_and_B - product

print("Formula:")

print(
    "P(A and B) = P(A) × P(B)"
)

print("\nObserved P(A and B):")

print(
    f"{P_A_and_B:.6f}"
)

print("\nP(A) × P(B):")

print(
    f"{P_A:.6f} × {P_B:.6f} = {product:.6f}"
)

print("\nDifference:")

print(
    f"{difference:.6f}"
)

if np.isclose(
    P_A_and_B,
    product
):

    print(
        "\nConclusion: A and B appear to be independent."
    )

else:

    print(
        "\nConclusion: A and B do NOT appear to be independent."
    )

    print(
        "Because P(A and B) is different from "
        "P(A) × P(B), the two events show an "
        "association in this sample."
    )


# ============================================================
# 17. BAYES' THEOREM
# ============================================================

print("\n" + "=" * 70)
print("17. BAYES' THEOREM")
print("=" * 70)

# P(not A)

P_not_A = 1 - P_A

# P(B | not A)

not_A = ~A

P_B_given_not_A = (
    (B & not_A).sum()
    / not_A.sum()
)

print(
    "P(A):",
    round(P_A, 6)
)

print(
    "P(B|A):",
    round(P_B_given_A, 6)
)

print(
    "P(not A):",
    round(P_not_A, 6)
)

print(
    "P(B|not A):",
    round(P_B_given_not_A, 6)
)


# ============================================================
# 18. APPLY BAYES' FORMULA
# ============================================================

print("\n" + "=" * 70)
print("18. BAYES CALCULATION")
print("=" * 70)

# Bayes formula:
#
# P(A|B) =
#
# [P(B|A) × P(A)]
#
# -------------------------------------------
#
# [P(B|A) × P(A)]
# +
# [P(B|not A) × P(not A)]

numerator = (
    P_B_given_A * P_A
)

denominator = (
    (P_B_given_A * P_A)
    +
    (P_B_given_not_A * P_not_A)
)

Bayes_P_A_given_B = (
    numerator / denominator
)

print("Formula:")

print(
    "P(A|B) = [P(B|A) × P(A)] / "
    "[P(B|A) × P(A) + "
    "P(B|not A) × P(not A)]"
)

print("\nSubstitution:")

print(
    f"P(A|B) = "
    f"[{P_B_given_A:.6f} × {P_A:.6f}] / "
    f"[({P_B_given_A:.6f} × {P_A:.6f}) + "
    f"({P_B_given_not_A:.6f} × {P_not_A:.6f})]"
)

print(
    f"\nBayes P(A|B) = "
    f"{Bayes_P_A_given_B:.6f}"
)


# ============================================================
# 19. VERIFY BAYES WITH DIRECT CONDITIONAL PROBABILITY
# ============================================================

print("\n" + "=" * 70)
print("19. BAYES VERIFICATION")
print("=" * 70)

print(
    "Direct P(A|B):",
    round(P_A_given_B, 6)
)

print(
    "Bayes P(A|B):",
    round(Bayes_P_A_given_B, 6)
)

bayes_difference = abs(
    P_A_given_B -
    Bayes_P_A_given_B
)

print(
    "Absolute difference:",
    bayes_difference
)

if np.isclose(
    P_A_given_B,
    Bayes_P_A_given_B
):

    print(
        "\nConclusion: Bayes' theorem agrees with "
        "the direct conditional-probability calculation."
    )

else:

    print(
        "\nConclusion: The two calculations do not agree."
    )


# ============================================================
# 20. NORMAL DISTRIBUTION
# ACADEMIC READINESS SCORE
# ============================================================

print("\n" + "=" * 70)
print("20. NORMAL DISTRIBUTION ANALYSIS")
print("=" * 70)

academic_scores = df[
    "Academic_Readiness_Score"
]

mean_academic = academic_scores.mean()

std_academic = academic_scores.std()

print(
    "Mean Academic Readiness Score:",
    round(mean_academic, 4)
)

print(
    "Standard Deviation:",
    round(std_academic, 4)
)


# ============================================================
# 21. FIND HIGHEST AND LOWEST SCORES
# ============================================================

print("\n" + "=" * 70)
print("21. HIGHEST AND LOWEST ACADEMIC READINESS SCORES")
print("=" * 70)

highest_index = academic_scores.idxmax()

lowest_index = academic_scores.idxmin()

highest_score = academic_scores.loc[
    highest_index
]

lowest_score = academic_scores.loc[
    lowest_index
]

highest_student_id = df.loc[
    highest_index,
    "Student_ID"
]

lowest_student_id = df.loc[
    lowest_index,
    "Student_ID"
]

print(
    "Highest Academic Readiness Score:",
    highest_score
)

print(
    "Student ID:",
    highest_student_id
)

print(
    "\nLowest Academic Readiness Score:",
    lowest_score
)

print(
    "Student ID:",
    lowest_student_id
)


# ============================================================
# 22. CALCULATE Z-SCORES
# ============================================================

print("\n" + "=" * 70)
print("22. Z-SCORES")
print("=" * 70)

# Formula:
#
# z = (x - mean) / standard deviation

z_highest = (
    highest_score -
    mean_academic
) / std_academic

z_lowest = (
    lowest_score -
    mean_academic
) / std_academic

print(
    "Z-score formula:"
)

print(
    "z = (x - mean) / standard deviation"
)

print("\nHighest score Z-score:")

print(
    f"z = ({highest_score} - "
    f"{mean_academic:.4f}) / "
    f"{std_academic:.4f}"
)

print(
    f"z = {z_highest:.4f}"
)

print("\nLowest score Z-score:")

print(
    f"z = ({lowest_score} - "
    f"{mean_academic:.4f}) / "
    f"{std_academic:.4f}"
)

print(
    f"z = {z_lowest:.4f}"
)


# ============================================================
# 23. INTERPRET Z-SCORES
# ============================================================

print("\n" + "=" * 70)
print("23. Z-SCORE INTERPRETATION")
print("=" * 70)

print(
    f"The highest Academic Readiness Score "
    f"({highest_score}) has a Z-score of "
    f"{z_highest:.4f}."
)

print(
    f"This means the highest score is approximately "
    f"{abs(z_highest):.2f} standard deviations "
    f"ABOVE the mean."
)

print(
    f"\nThe lowest Academic Readiness Score "
    f"({lowest_score}) has a Z-score of "
    f"{z_lowest:.4f}."
)

print(
    f"This means the lowest score is approximately "
    f"{abs(z_lowest):.2f} standard deviations "
    f"BELOW the mean."
)


# ============================================================
# 24. 68-95-99.7 EMPIRICAL RULE
# ============================================================

print("\n" + "=" * 70)
print("24. 68-95-99.7 EMPIRICAL RULE")
print("=" * 70)

print(
    "For an approximately normal distribution:"
)

print(
    "\nWithin ±1 standard deviation:"
)

print(
    "Expected percentage = 68%"
)

print(
    "\nWithin ±2 standard deviations:"
)

print(
    "Expected percentage = 95%"
)

print(
    "\nWithin ±3 standard deviations:"
)

print(
    "Expected percentage = 99.7%"
)


# ============================================================
# 25. EMPIRICAL RULE INTERVALS
# ============================================================

print("\n" + "=" * 70)
print("25. EMPIRICAL RULE INTERVALS")
print("=" * 70)

lower_1sd = (
    mean_academic -
    std_academic
)

upper_1sd = (
    mean_academic +
    std_academic
)

lower_2sd = (
    mean_academic -
    2 * std_academic
)

upper_2sd = (
    mean_academic +
    2 * std_academic
)

lower_3sd = (
    mean_academic -
    3 * std_academic
)

upper_3sd = (
    mean_academic +
    3 * std_academic
)

print(
    f"Mean ± 1 SD: "
    f"{lower_1sd:.2f} to "
    f"{upper_1sd:.2f}"
)

print(
    f"Mean ± 2 SD: "
    f"{lower_2sd:.2f} to "
    f"{upper_2sd:.2f}"
)

print(
    f"Mean ± 3 SD: "
    f"{lower_3sd:.2f} to "
    f"{upper_3sd:.2f}"
)


# ============================================================
# 26. HISTOGRAM
# ============================================================

print("\n" + "=" * 70)
print("26. ACADEMIC READINESS SCORE DISTRIBUTION")
print("=" * 70)

plt.figure(figsize=(10, 6))

plt.hist(
    academic_scores,
    bins=20,
    edgecolor="black"
)

plt.axvline(
    mean_academic,
    linestyle="--",
    linewidth=2,
    label="Mean"
)

plt.xlabel(
    "Academic Readiness Score"
)

plt.ylabel(
    "Number of Students"
)

plt.title(
    "Distribution of Academic Readiness Scores"
)

plt.legend()

plt.show()


# ============================================================
# 27. FIVE MEANINGFUL STATISTICAL OBSERVATIONS
# ============================================================

print("\n" + "=" * 70)
print("27. FIVE MEANINGFUL STATISTICAL OBSERVATIONS")
print("=" * 70)

print(
    f"""
1. Academic readiness has the greatest variability among the
   five main variables. It has the highest standard deviation
   ({greatest_std_value:.4f}), as well as the largest range and IQR.

2. Outliers can influence the mean. For Weekly Study Hours,
   the mean changed from {before_mean:.4f} before outlier removal
   to {after_mean:.4f} after removing IQR-detected outliers.
   The median changed from {before_median:.4f} to {after_median:.4f}.

3. Part-time employment and high stress do not appear to be
   independent. P(A and B) = {P_A_and_B:.4f}, whereas
   P(A) × P(B) = {product:.4f}. Since these values are different,
   the independence condition is not satisfied.

4. Bayes' theorem confirms the conditional probability result.
   Direct P(A|B) = {P_A_given_B:.4f}, while Bayes' theorem gives
   P(A|B) = {Bayes_P_A_given_B:.4f}. The values agree.

5. Academic Readiness Scores contain observations in the tails.
   The highest score has a Z-score of {z_highest:.4f}, while the
   lowest score has a Z-score of {z_lowest:.4f}. The lowest score
   is slightly more than three standard deviations below the mean.
"""
)


# ============================================================
# 28. FINAL CONCLUSION
# ============================================================

print("\n" + "=" * 70)
print("28. FINAL CONCLUSION")
print("=" * 70)

print(
"""
The Student Wellbeing Survey was successfully analyzed using
descriptive statistics, measures of variability, IQR-based
outlier detection, probability, conditional probability,
mutual exclusivity, independence testing, Bayes' theorem,
Z-scores, and the 68-95-99.7 empirical rule.

The analysis provides numerical results and interpretations
for all requested parts of the assignment.
"""
)


# ============================================================
# END OF ASSIGNMENT
# ============================================================

print("\n" + "=" * 70)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 70)
```
